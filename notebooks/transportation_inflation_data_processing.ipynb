```python
# Import pandas for reading, cleaning, and processing the CPI dataset.
import pandas as pd
```

```python
# Load the Transportation CPI CSV file. 
file_path = "../data/raw/Consumer Price Index for All Urban Consumers Transportation in U.S. City Average.csv"

df = pd.read_csv(file_path)
```

```python
# Preview the dataset structure.
print(df.head())
print(df.info())
```

```python
# Convert the observation_date column from text into a real date format.
df["observation_date"] = pd.to_datetime(df["observation_date"])
```

```python
# Rename the CPI value column to something easier to understand.
df = df.rename(columns={
    "CPITRNSL": "transportation_cpi"
})
```

```python
# Filter the dataset so it starts from January 1, 2024.
filtered_df = df[df["observation_date"] >= "2024-01-01"].copy()
```

```python
# Sort the filtered data by date.
filtered_df = filtered_df.sort_values("observation_date")
```

```python
# Calculate month over month percent change.
filtered_df["mom_percent_change"] = filtered_df["transportation_cpi"].pct_change(fill_method=None) * 100
```

```python
# Calculate year over year percent change.
filtered_df["yoy_percent_change"] = filtered_df["transportation_cpi"].pct_change(periods=12, fill_method=None) * 100
```

```python
# Add separate year and month columns.
filtered_df["year"] = filtered_df["observation_date"].dt.year
filtered_df["month"] = filtered_df["observation_date"].dt.month
filtered_df["month_name"] = filtered_df["observation_date"].dt.strftime("%b")
```

```python
# Check the final processed dataset.
print(filtered_df.head())
print(filtered_df.tail())
```

```python
# Save the processed dataset as a new CSV file.
output_path = "../data/processed/transportation_cpi_2024_2026_processed.csv"

filtered_df.to_csv(output_path, index=False)
```

```python
# Identify the largest monthly increases.
# This helps find possible peak months for visualization and news comparison.
top_mom_peaks = filtered_df.sort_values(
    "mom_percent_change",
    ascending=False
).head(5)

print("Top month over month Transportation CPI increases:")
print(top_mom_peaks[[
    "observation_date",
    "transportation_cpi",
    "mom_percent_change",
    "yoy_percent_change"
]])
```

```python
# Optional: Identify the largest year over year increases.
# This helps show which months had the strongest inflation compared with the same month one year earlier.
top_yoy_peaks = filtered_df.sort_values(
    "yoy_percent_change",
    ascending=False
).head(5)

print("Top year over year Transportation CPI increases:")
print(top_yoy_peaks[[
    "observation_date",
    "transportation_cpi",
    "mom_percent_change",
    "yoy_percent_change"
]])
```
